```python
from openai import OpenAI

BASE_URL = "http://localhost:11434/v1"
API_KEY = "ollama"
MODEL = "qwen2.5:7b"
LLM_OPTIONS = {"reasoning_effort": "none"} if API_KEY == "ollama" else {}
client = OpenAI(base_url=BASE_URL, api_key=API_KEY)
```

```python
def ask_llm(prompt: str) -> str:
    """
    Ask the LLM a question and return the response.
    """
    r = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "user", "content": prompt},
        ],
        **LLM_OPTIONS)
    return r.choices[0].message.content
```

```python
prompt = "What is the capital of France? tell me by 1 word"
ask_llm(prompt)
```

```python
prompt = "What is 384712 * 9931? Just compute it directly"
print(ask_llm(prompt))
print("real answer is str(384712 * 9931) = ", str(384712 * 9931))
```

```python
prompt = "What did I just ask you in my previous message?"
print(ask_llm(prompt))
```
