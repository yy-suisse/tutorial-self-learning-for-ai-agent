```python
from agentlib.llm import get_client

#client, model = get_client()                      # uses LLM_PROFILE from .env
# client, model = get_client("openrouter_free")   # or force one profile
client, model = get_client(require_local=True)  # use this in every HERO project
```

```python

r = client.chat.completions.create(
    model=model,
    messages=[{"role": "system", "content": "please write a query for PubMed with the following description. maybe include MeSH terms and boolean operators. The query should be specific to the topic and return relevant results."},
              {"role": "user", "content": "find me the papers that describe different ways we use robots for children mental health followup, tracking, analysis and problem intervention"},],
    temperature=0.5,
)
print(r.choices[0].message.content)
print(r.usage)
```
